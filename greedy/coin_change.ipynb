```python
class Solution:

    def coin_change_greedy(self, coins, amount):
        coins = sorted(coins, reverse = True)
        total = 0
        


        for coin in coins:

            val = amount // coin
            amount = amount - (val*coin)
            total +=val

            if amount ==0:
                return total
        return -1
```

```python
coins = [500,100,10,1]


sol = Solution()
sol.coin_change_greedy(coins, 3240)


```
