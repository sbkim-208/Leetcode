```python
intervals = [[1,2],[2,3],[3,4],[1,3]]


intervals.sort(key = lambda x:x[1])
intervals

for interval in intervals:
    print(interval[0])

choice = intervals[0]
choice[1]
```

```python
class Solution:

    def interval_schedule(self, intervals):

        intervals.sort(key = lambda x:x[1])


        choice = intervals[0]
        count = 1


        for interval in intervals[1:]:

            if interval[0] >= choice[1]:
                choice = interval
                count +=1

        return count 








        
```

```python
sol = Solution()

sol.interval_schedule(intervals)
```
