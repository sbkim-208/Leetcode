```python
def findRedundantConnection(edges):
    n = len(edges)
    parent = list(range(n + 1))  # 노드 번호는 1부터 시작

    def find(x):
        while parent[x] != x:
            parent[x] = parent[parent[x]]  # 경로 압축
            x = parent[x]
        return x

    def union(x, y):
        rootX, rootY = find(x), find(y)
        if rootX == rootY:
            return False  # 이미 연결되어 있음 -> 사이클
        parent[rootX] = rootY
        return True

    for u, v in edges:
        if not union(u, v):
            return [u, v]

    return []
```
