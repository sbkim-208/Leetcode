```python
from collections import defaultdict


class Solution:
    def schedule(self, n,prerequisites):

        graph = defaultdict(list)


        for aft,bef in prerequisites:
            graph[bef].append(aft)

        visited,need_visit = set(), set() 
        def dfs(key):

            if key in need_visit:
                return False
            if key in visited:
                return True
            
            need_visit.add(key)


            
            for neighbor in graph[key]:

                if not dfs(neighbor):
                    return False
                
            
            need_visit.remove(key)
            visited.add(key)

            return True

        



        for i in range(n):
            if not dfs(i):
                return False
        return True
            
        
        

        

```

```python
prerequisites = [[1,0],[2,1],[3,2],[4,3]]
numCourses = 5

sol = Solution()
sol.schedule(numCourses, prerequisites)
```
