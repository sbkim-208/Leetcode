```python

 

class Solution:

    def islands(grid):

        count = 0
        visited = set()


        def dfs(i,j):


            if i < 0 or i >=len(grid):
                return 
            if j < 0 or j >=len(grid[0]):
                return
            
            if grid[i][j] == "0":
                return 
            
            if (i,j) in visited:
                return 
            
            visited.add((i,j))

            dfs(i,j+1)
            dfs(i,j-1)
            dfs(i-1,j)
            dfs(i+1,j)

            
        

        for i in range(len(grid)):
            for j in range(len(grid[0])):

                if grid[i][j] != "0" and (i,j) not in visited:

                    dfs(i,j)
                count +=1
            






        return count



```
