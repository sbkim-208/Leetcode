```python
from typing import List
from collections import deque

class Solution:
    def orangesRotting(self, grid: List[List[int]]) -> int:
        queue = deque()
        row = len(grid)
        col = len(grid[0])
        fresh = 0

        # 썩은 오렌지 찾기 + 신선한 오렌지 개수 세기
        for i in range(row):
            for j in range(col):
                if grid[i][j] == 2:
                    queue.append((i, j))
                elif grid[i][j] == 1:
                    fresh += 1

        if fresh == 0:
            return 0   # 처음부터 신선한 오렌지가 없으면 0분

        minutes = 0
        directions = [(1,0), (-1,0), (0,1), (0,-1)]

        while queue and fresh > 0:
            minutes += 1
            for _ in range(len(queue)):
                ni, nj = queue.popleft()

                for di, dj in directions:
                    ni2, nj2 = ni + di, nj + dj

                    if 0 <= ni2 < row and 0 <= nj2 < col and grid[ni2][nj2] == 1:
                        grid[ni2][nj2] = 2      # 썩은 상태로 표시
                        fresh -= 1               # 신선한 오렌지 개수 감소
                        queue.append((ni2, nj2))

        return minutes if fresh == 0 else -1 #  다썩는거 불가능
```
