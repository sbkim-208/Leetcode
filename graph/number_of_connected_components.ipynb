```python
from typing import List
from collections import defaultdict

class Solution:
    def countComponents(self, n: int, edges: List[List[int]]) -> int:

        adj = defaultdict(list)

        for a,b in edges: 
            adj[a].append(b)
            adj[b].append(a)
        
        visited = set()
        count = 0


        def dfs(num): 

            if num in visited: 
                return 

            visited.add(num)

            for key in adj[num]: 
                dfs(key)




        for i in range(n): 
            if i not in visited: 
                dfs(i)
                count +=1

        return count 
```
