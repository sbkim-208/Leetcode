```python
from typing import List
from collections import defaultdict

class Solution:
    def countComponents(self, n: int, edges: List[List[int]]) -> int:

        adj = defaultdict(list)

        for a,b in edges: 
            adj[a].append(b)
            adj[b].append(a)
        
        visited = set()
        count = 0


        def dfs(num): 

            if num in visited: 
                return 

            visited.add(num)

            for key in adj[num]: 
                dfs(key)




        for i in range(n): 
            if i not in visited: 
                dfs(i)
                count +=1

        return count 
```

```python
## Union and Find 방식
# O(E * a(V)) = O(E)



class Solution: 
    def number_of_connected(self, n, edges): 
        self.parent = list(range(n))# 0부터 시작함 
        self.count = n
        for u,v in edges: 
            self.union(u,v)
        return self.count
        

    
    def find(self, x):
        if self.parent[x] != x: 
            self.parent[x] = self.find(self.parent[x])
        

        return self.parent[x] 
    
    def union(self, x,y):
        rootX, rootY = self.find(x), self.find(y)

        if rootX == rootY: 
            return  #  같으면 냅둠 
        
        self.parent[rootY] = rootX

        self.count -=1 
    
    


```
