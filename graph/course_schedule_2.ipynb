```python
def courses_2(numbercourses, prerequisites): 

    pre = defaultdict(list)
    visited = set()
    need_visit = set()
    result = []
    for a, b in prerequisites: 
        pre[b].append(a)

    def dfs(num): 

    

        if num in need_visit: 
            return False
        if num in visited: 
            return True
        need_visit.add(num)

        

       

        for  val in pre[num]: 

           if not dfs(val): 
               return False
        
        need_visit.remove(num)
        visited.add(num)
        result.append(num)
        return True 





    for i in range(numbercourses): 
        if not dfs(i):  # false일경우에 return false
            return []


    
    return result
```

```python
from collections import defaultdict

# 테스트 1: 기본 예시
print(courses_2(2, [[1,0]]))
# 기대값: [0, 1]

# 테스트 2: 여러 경로
print(courses_2(4, [[1,0],[2,0],[3,1],[3,2]]))
# 기대값: [0,1,2,3] 또는 [0,2,1,3]

# 테스트 3: 사이클 → 빈 리스트
print(courses_2(2, [[1,0],[0,1]]))
# 기대값: []

# 테스트 4: prerequisites 없음
print(courses_2(3, []))
# 기대값: [0,1,2]

# 테스트 5: 고립된 강의 있음
print(courses_2(4, [[1,0]]))
# 기대값: [0,1,2,3] (2,3은 순서 무관)
```
