```python
from collections import defaultdict
import heapq
from typing import List

class Solution:
    def networkDelayTime(self, times: List[List[int]], n: int, k: int) -> int:
        visited = {} # set을 사용해서, 바로 접근해서 코드를 간결화시킴
        graph = defaultdict(list)

        for u, v, w in times:
            graph[u].append((v, w))

        queue = [(0, k)]

        while queue:
            cost, node = heapq.heappop(queue)
            if node in visited:
                continue
            visited[node] = cost

            for neighbor, weight in graph[node]:
                # 자동으로 최솟값이 앞으로 정렬됨
                heapq.heappush(queue, (cost + weight, neighbor))

        if len(visited) == n:
            return max(visited.values())
        else:
            return -1
        #visited 방식은 힙에서 꺼내는 순간 "이 노드의 최단거리는 확정" 이라는 다익스트라 핵심 원리를 그대로 활용
        #힙에서 꺼냄 → 최솟값 보장 → 즉시 확정 → visited에 추가
```
