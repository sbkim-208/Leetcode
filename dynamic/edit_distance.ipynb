```python
class Solution:
    def minDistance(self, word1: str, word2: str) -> int:
        m, n = len(word1), len(word2)
        dp = [[0] * (n + 1) for _ in range(m + 1)]

        # base case: 한쪽이 빈 문자열일 때
        for i in range(m + 1):
            dp[i][0] = i   # word1의 i글자를 전부 삭제해야 빈 문자열이 됨
        for j in range(n + 1):
            dp[0][j] = j   # word2의 j글자를 전부 삽입해야 word2가 됨

        for i in range(1, m + 1):
            for j in range(1, n + 1):
                if word1[i-1] == word2[j-1]:
                    dp[i][j] = dp[i-1][j-1]        # 이미 같으면 연산 불필요
                else:
                    dp[i][j] = 1 + min(
                        dp[i][j-1],      # insert
                        dp[i-1][j],      # delete
                        dp[i-1][j-1]     # replace
                    )

        return dp[m][n]
```
