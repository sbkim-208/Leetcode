```python
from typing import List

class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:

        if not nums:
            return 0
        n = len(nums)
        dp = [1] *(n)
        dp[0] = 1

        for i in range(1,n): 
            for j in range(i): 
                
                if nums[j]<nums[i]:
                    dp[i] = max(dp[i], dp[j] + 1)
        return max(dp)
```

```python
nums = [10, 9, 2, 5, 3, 7, 101, 18,102]
sol = Solution()
sol.lengthOfLIS(nums)
```
