```python
class Solution:
    def numDecodings(self, s: str) -> int:
        if not s or s[0] == '0':
            return 0

        n = len(s)
        # dp[i] = s[0:i]까지 디코딩하는 방법의 수
        dp = [0] * (n + 1)
        dp[0] = 1  # 빈 문자열: 방법 1가지 (아무것도 안 함)
        dp[1] = 1  # 첫 글자는 '0'이 아니므로 항상 1가지

        for i in range(2, n + 1):
            # 마지막 한 글자만 떼서 확인 (s[i-1])
            one_digit = int(s[i-1])
            if one_digit >= 1:  # '1'~'9'는 유효
                dp[i] += dp[i-1]

            # 마지막 두 글자를 떼서 확인 (s[i-2:i])
            two_digit = int(s[i-2:i])
            if 10 <= two_digit <= 26:  # '10'~'26'만 유효
                dp[i] += dp[i-2]

        return dp[n]
```
