```python
def canPartition(nums):
    total = sum(nums)
    
    # 합이 홀수면 절대 두 부분집합으로 똑같이 나눌 수 없음
    if total % 2 != 0:
        return False
    
    target = total // 2
    n = len(nums)
    
    # dp[i] = target 중 i를 만들 수 있는지 여부
    dp = [False] * (target + 1)
    dp[0] = True  # 아무것도 선택 안 하면 0은 항상 만들 수 있음
    
    for num in nums:
        # 뒤에서부터 순회해야 같은 원소를 중복 사용하지 않음 (0/1 배낭)
        for j in range(target, num - 1, -1):
            dp[j] = dp[j] or dp[j - num]
    
    return dp[target]


# 테스트
print(canPartition([1, 5, 11, 5]))   # True  -> [1,5,5]와 [11]
print(canPartition([1, 2, 3, 5]))    # False -> 11로 나눌 수 없음
```
