```python
class Solution:

    def coin_change2(self, coins, amount):
        

        dp = (amount +1) *[0]
        dp[0] = 1

        for coin in coins: 
            for i in range(coin, amount +1): 
                dp[i] += dp[i-coin]

        return dp[amount]

        

```

```python

coins = [1,2,5]
sol = Solution()
sol.coin_change2(coins, 5)
```
