```python
def maxProduct(nums):
    max_dp = min_dp = result = nums[0]
    
    for i in range(1, len(nums)):
        curr = nums[i]
        # 갱신 전에 max_dp를 미리 저장 (min_dp 계산 시 필요)
        candidates = (curr, max_dp * curr, min_dp * curr)
        max_dp = max(candidates)
        min_dp = min(candidates)
        
        result = max(result, max_dp)
    
    return result
```
