```python
from typing import List
```

```python
class Solution:
    def maxProduct(self, nums: List[int]) -> int:

        n = len(nums)
        max_dp = min_dp = result = nums[0]

        for i in range(1, n): 
            old_min = min_dp
            old_max = max_dp

            min_dp = min(old_min, nums[i]*old_max, nums[i]*max_dp)
            max_dp = max(old_max, nums[i]*old_min, nums[i]*max_dp)

            result = max(result, max_dp)

        return result 
```

```python

nums = [-2,3,-2]
sol = Solution()
sol.maxProduct(nums)
```
