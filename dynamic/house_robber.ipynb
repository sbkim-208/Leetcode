```python
class Solution:

    def house_robber(self,nums):
        if len(nums) ==1:
            return nums[0]

        dp = [0] * len(nums)

        dp[0] = nums[0]
        dp[1] = max(dp[0], nums[1])

        for i in range(2,len(nums)):
            dp[i] = max(dp[i-2]+nums[i], dp[i-1])
        

        return dp[len(nums)-1]

```

```python
nums = [2,7,9,3,1]

sol = Solution()
sol.house_robber(nums)
```
