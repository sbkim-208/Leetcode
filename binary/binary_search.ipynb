```python
class Solution: 
    def binary_search(self, arr, target): 

        left, right = 0, len(arr) -1

        while left <= right: 

            mid = (left + right) //2

            if arr[mid] == target:
                return True
            if arr[mid] < target: 
                left = mid+1
            if arr[mid] > target:
                right = mid -1
            
        return False
```

```python
arr = [1,3,5,7]
sol  = Solution()

sol.binary_search(arr, 7)
```
