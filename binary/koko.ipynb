```python
import math

def koko(piles, h): 
    left , right = 0, max(piles)

    

    while left <= right: 
        mid = (left + right) //2
        total = 0

        for pile in piles: 
            total += math.ceil(pile/mid)

        if total <= h: 
            right = mid -1 


        
        if total > h: 

            left = mid + 1

    return left  # 마지막으로 가능한 가장 작은 갑4ㅅ 
    








    
```
