```python
def search(matrix, target): 


   left = 0
   right = len(matrix) * len(matrix[0]) -1

   while left <=right:
      mid = (left + right) //2

      row = mid // (len(matrix[0])) # 여러 조합으로 풀기 

      col = mid % (len(matrix[0]))

      if target == matrix[row][col]:
         return True


      if target > matrix[row][col]:
          left = mid + 1
      

      if target < matrix[row][col]:
         right = mid -1
   return False
    



      
      


    
```
