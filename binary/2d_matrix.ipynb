```python
class Solution: 

    def search_2d(self, matrix, target): 


        left, right = 0, len(matrix[0]) * len(matrix) -1

        while left <= right : 

            mid = (left + right)//2

            row = mid // len(matrix[0])
            col = mid % len(matrix[0])

            if target  == matrix[row][col]: 
                return True
            if target < matrix[row][col]: 
                right  = mid -1
            if target > matrix[row][col] : 
                left = mid + 1
        return False
```

```python
matrix = [
  [1,  3,  5,  7],
  [10, 11, 16, 20],
  [23, 30, 34, 60]
]
target = 13
```

```python
sol = Solution()



sol.search_2d(matrix, target)
```
