```python
class Solution:

    def min_number(self, nums):


        left, right = 0, len(nums)-1

        while left < right:
            
            mid = (right+ left) //2

            if nums[mid] > nums[right]:
                left = mid + 1

            else:
                right = mid 
            
        
        return nums[left]
            
```

```python

nums =[3, 1, 2]
sol = Solution()

sol.min_number(nums)
```
