```python
from collections  import defaultdict

class TimeMap: 
    def __init__(self):
        self.store = dict()
# self.store = dict의 경부 내부생성 -> 외부에서 값 안받음 


    def set(self, key,value, timestamp): 


        if key not in self.store: 

            self.store[key] = []
        
        self.store[key].append((value, timestamp))
    
    def get(self, key, time):

        if not key in self.store: 
            return None
        
        

        pairs = self.store[key]
        left, right = 0, len(pairs) -1 # index값을 나타냄 


        while left <= right: 
            mid = (left + right) //2 # 

            val, ts = pairs[mid]

            if ts <= time: 
                left = mid +1
            
            else : 
                right = mid -1
        
       
        if right < 0: 
            return ""
        return pairs[right][0]


           

       
        





```
