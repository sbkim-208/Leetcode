```python
def lowest_common_ancestor(root, p, q):

    def dfs(node):
        if not node:
            return None
        if node == p or node == q:
            return node

        left = dfs(node.left)
        right = dfs(node.right)

        if left and right:
            return node
        return left or right

    return dfs(root) # 안에서 찾아야함 
```
