```python
class Solution: 
    def count_good_nodes(self, root):


        def dfs(node, max_node):
            if not node: 
                return 0
            new_max = max(node.val, max_node)
            if node.val >= max_node:  # if node.val >= new_max
                count = 1
            else: 
                count = 0


            return count + dfs(node.left, new_max) + dfs(node.right, new_max)




        return dfs(root, max_node=float("-inf"))
```
