```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
class Solution:
    def recoverTree(self, root):
        self.first = None
        self.second = None
        self.prev = None




        def inorder(node): 

            if not node: return 

            inorder(node.left)

           
            if self.prev and self.prev.val > node.val: 
                if not self.first: 
                    self.first = self.prev
                
                self.second = node
            self.prev = node

            inorder(node.right)
        
        inorder(root)

        self.first.val, self.second.val = self.second.val, self.first.val

        return root 
```

```python
# 테스트 케이스 1: 인접 swap [1,3,null,null,2]
root1 = TreeNode(1)
root1.left = TreeNode(3)
root1.left.right = TreeNode(2)

# 테스트 케이스 2: 먼 거리 swap [3,1,4,null,null,2]
root2 = TreeNode(3)
root2.left = TreeNode(1)
root2.right = TreeNode(4)
root2.right.left = TreeNode(2)

def inorder_print(root):
    if not root: return
    inorder_print(root.left)
    print(root.val, end=' ')
    inorder_print(root.right)

s = Solution()

print("케이스 1 before:", end=' '); inorder_print(root1)
s.recoverTree(root1)
print("\n케이스 1 after: ", end=' '); inorder_print(root1)

print("\n\n케이스 2 before:", end=' '); inorder_print(root2)
s.recoverTree(root2)
print("\n케이스 2 after: ", end=' '); inorder_print(root2)
```
