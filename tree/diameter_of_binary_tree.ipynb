```python
class Solution: 
    def max_diameter(self, root):
        self.diameter = 0




        def dfs(node):
            if node is None:
               return 0

            left = dfs(node.left)
            right = dfs(node.right)

            self.diameter = max(left + right, self.diameter)


            return max(left, right) +1
        
        dfs(root)
        return self.diameter
```
