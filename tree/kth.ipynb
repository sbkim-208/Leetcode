```python
def validate_bianry_search_tree( root, min_value = float("-inf"), max_value = float("inf")):
    if root is None:                    # ✅ None 체크
        return True

    if root.val <= min_value or root.val >= max_value: # 안되는걸 쓸것 
        return False
    
    return validate_bianry_search_tree(root.left,min_value ,root.val) and validate_bianry_search_tree(root.right, root.val, max_value)



```
