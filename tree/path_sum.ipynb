```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def hasPathSum(self, root: Optional[TreeNode], targetSum: int) -> bool:

        if root is None: 
            return False

        rest_sum = targetSum - root.val

        if root.left is None and root.right is None and rest_sum == 0 : 
            return True

        




        return self.hasPathSum(root.left, rest_sum) or self.hasPathSum(root.right, rest_sum)
        
```
