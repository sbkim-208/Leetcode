```python
class Solution:
    def invert_tree_pre_order(self, root):



        if root is None:
            return None
        

        root.left, root.right = root.right, root.left

        self.invert_tree_pre_order(root.left)
        self.invert_tree_pre_order(root.right)

        return root 
```

```python
class Solution:
    def invert_tree_post_order(self,root):
        if root is None:
            return None
        
        self.invert_tree_post_order(root.left)
        self.invert_tree_post_order(root.right)

        root.left, root.right = root.right, root.left

        return root
```
