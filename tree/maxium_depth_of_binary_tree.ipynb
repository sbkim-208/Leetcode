```python
class Solution:

    def maxium_depth_of_binary_tree(self, root):

      
        if root is None:
                return 0
            
        left = self.maxium_depth_of_binary_tree(root.left)
        right = self.maxium_depth_of_binary_tree(root.right)
            

        return max(right, left) + 1
        


        



```

```python
from collections import deque

class Solution:

    def bfs(self, root):

        if root is None:
            return 0 ## 예외처리

        queue = deque([root]) # 첫 시작 
        depth = 0

        while queue: 

            depth +=1

            

            for _ in range(len(queue)):
                node = queue.popleft()
                if node.left:
                    queue.append(node.left)
                
                if node.right :
                    queue.append(node.right)
        return depth 
                
```
