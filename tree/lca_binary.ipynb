BST → 값 비교로 방향 앎
한쪽만 탐색!

가운데 찾기


## Time complexity
    - O(log n) : 절반씩 줄여가기
    - 재귀 스택 트리 높이만큼


```python
class Solution: 

    def binary_search(self, root, p, q): 

        def dfs(node): 
            if not node: 
                return None
            if p.val <= node.val and q.val <= node.val: 
                return dfs(node.left)
            if p.val > node.val and q.val > node.val: 
                return dfs(node.right)
            return node

        return dfs(root)
```
