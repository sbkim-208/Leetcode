```python
class Solution:
    def sumNumbers(self, root) -> int:

        cur = 0

     


        def dfs(node, cur):

            if node is None: 
                return 0

            cur = cur * 10 + node.val

            if node.left is None and node.right is None: 
                return cur

            return dfs(node.left, cur) + dfs(node.right, cur)
        return dfs(root, cur)

    

            
```
