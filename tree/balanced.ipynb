```python
class Solution:

    def balanced_binary_tree(self, root):

        
        

        def dfs(node):
            if not node :
               return 0

            

            left = dfs(node.left) 
            right = dfs(node.right)
            if left == -1 or right == -1:
               return -1
            if abs(left - right) > 1:
                return -1


            
            return max(left, right) + 1
    
       
        return dfs(root) != -1
```
