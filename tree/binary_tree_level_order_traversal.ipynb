```python
from collections import deque


class Solution: 

    def traversal(self, root): 

        if root is None: 
            return []
        

        queue = deque([root])
        res = []
        

        while queue: 

            level = [] # while 시작하자마자 초기화

            for _ in range(len(queue)): 

                node = queue.popleft()

                level.append(node.val)


                if node.left: 
                    queue.append(node.left)
                if node.right: 
                    queue.append(node.right)
            res.append(level) # 해당 레벨의 node 추가 

        return res
                
```
