```python
def isvalidBST(self, root, min_val = float("-inf"), max_val = float("inf")):

    if not root:
        return True
    
    if root.val <= min_val or root.val>=max_val:
        return False
    
    return isvalidBST(root.left, min_val, root.val) and isvalidBST(root.right, root.val, max_val)
    
```
