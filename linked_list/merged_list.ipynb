```python
class Node:
    def __init__(self, val = 0, next = None):
        self.val = val
        self.next = next


class Solution:

    def merged(self,num1, num2):

        dummy = Node(0)

        curr = dummy


        while num1 and num2:

            if num1.val < num2.val:

                curr.next = num1
                num1 = num1.next

            else:
                curr.next = num2
                num2 = num2.next
            
            curr = curr.next


        if num1:
            curr.next = num1
        else:
            curr.next = num2

        return dummy.next
    def print_list(self):
        curr = self.head
        while curr:
            print(curr.val, end=" -> ")
            curr = curr.next
        print("None")
```

```python

# 테스트
sol = Solution()

# list1: 1 -> 2 -> 4
sol.head = Node(1)
sol.head.next = Node(2)
sol.head.next.next = Node(4)
list1 = sol.head

# list2: 1 -> 3 -> 4
sol.head = Node(1)
sol.head.next = Node(3)
sol.head.next.next = Node(4)
list2 = sol.head

sol.head = sol.merged(list1, list2)
sol.print_list()  
```
