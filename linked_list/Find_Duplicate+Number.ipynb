```python
def findDuplicate(nums: list[int]) -> int:
    # 1단계: 느린 포인터, 빠른 포인터로 사이클 안의 한 점 찾기
    slow, fast = nums[0], nums[nums[0]]
    while slow != fast:
        slow = nums[slow]
        fast = nums[nums[fast]]
    
    # 2단계: 한쪽을 시작점(0)으로 되돌리고, 같은 속도로 이동시켜 만나는 지점 = 사이클 시작점
    slow2 = 0
    while slow != slow2:
        slow = nums[slow]
        slow2 = nums[slow2]
    
    return slow
```
