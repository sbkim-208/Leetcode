```python
class Node:
    def __init__(self, x, next=None, random=None):
        self.val = x
        self.next = next
        self.random = random
class Solution:
    def copyRandomList(self, head: 'Optional[Node]') -> 'Optional[Node]':

        if not head: 
            return None

        visited = {}

        curr = head

        while curr: 
            visited[curr] = Node(curr.val)
            curr = curr.next

        curr = head

        while curr:
            copy = visited[curr]
            copy.next = visited.get(curr.next)
            copy.random = visited.get(curr.random)
            curr = curr.next
        

        return visited[head]
```
