```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def reverseKGroup(self, head: ListNode, k: int) -> ListNode:
        # k개 노드가 남아있는지 확인
        node = head
        count = 0
        while node and count < k:
            node = node.next
            count += 1
        
        # k개 미만이면 그대로 반환
        if count < k:
            return head
        
        # k개를 뒤집기
        prev = None
        curr = head
        for _ in range(k):
            next_node = curr.next
            curr.next = prev
            prev = curr
            curr = next_node
        
        # head는 이제 그룹의 마지막 노드가 됨
        # 다음 그룹을 재귀적으로 처리한 결과와 연결
        head.next = self.reverseKGroup(curr, k)
        
        return prev  # prev가 새로운 head
```
