```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def reorderList(self, head: Optional[ListNode]) -> None:
        """
        Do not return anything, modify head in-place instead.
        """
        curr = head

        slow = head
        fast = head

        while fast and fast.next: 
            slow = slow.next
            fast = fast.next.next

        second = slow.next
        slow.next = None # 끊어줌 

        # second 뒤집기
        prev = None
        curr = second

        while curr: 
            next_node = curr.next
            curr.next = prev
            prev = curr
            curr = next_node
        second = prev 

        # head와 second 병합 
        first = head

        while second: 
            first_next = first.next
            second_next = second.next

            first.next = second
            second.next = first_next
            first = first_next
            second = second_next

        return 

    def print_list(self, head):
        curr = head
        while curr:
            print(curr.val, end=" -> ")
            curr = curr.next
        print("None")


# ==================== 검증 코드 ====================

def build_list(values):
    dummy = ListNode()
    curr = dummy
    for v in values:
        curr.next = ListNode(v)
        curr = curr.next
    return dummy.next


sol = Solution()

print("Test 1 (odd, n=5):")
head1 = build_list([1, 2, 3, 4, 5])
sol.reorderList(head1)
sol.print_list(head1)
print("Expected: 1 -> 5 -> 2 -> 4 -> 3 -> None\n")

print("Test 2 (even, n=4):")
head2 = build_list([1, 2, 3, 4])
sol.reorderList(head2)
sol.print_list(head2)
print("Expected: 1 -> 4 -> 2 -> 3 -> None\n")

print("Test 3 (even, n=6):")
head3 = build_list([1, 2, 3, 4, 5, 6])
sol.reorderList(head3)
sol.print_list(head3)
print("Expected: 1 -> 6 -> 2 -> 5 -> 3 -> 4 -> None\n")

print("Test 4 (single node):")
head4 = build_list([1])
sol.reorderList(head4)
sol.print_list(head4)
print("Expected: 1 -> None\n")

print("Test 5 (two nodes):")
head5 = build_list([1, 2])
sol.reorderList(head5)
sol.print_list(head5)
print("Expected: 1 -> 2 -> None\n")
```
