## Product of Array Except Self

Given an integer array nums, return an array answer such that answer[i] is equal to the product of all the elements of nums except nums[i].

정수 배열 nums가 주어졌을 때, answer[i]가 nums[i]를 제외한 나머지 모든 원소들의 곱과 같은 배열 answer를 반환하세요.

The product of any prefix or suffix of nums is guaranteed to fit in a 32-bit integer.

nums의 어떤 prefix(앞부분) 또는 suffix(뒷부분)의 곱도 32비트 정수 범위 안에 들어감이 보장됩니다.

You must write an algorithm that runs in O(n) time and without using the division operation.

나눗셈 연산을 사용하지 않고 O(n) 시간에 동작하는 알고리즘을 작성해야 합니다.

- trapping rain water과 비슷

```python
class Solution:
    def productExceptSelf(self, nums: List[int]) -> List[int]:

        n = len(nums)

        answer = [1] * n
        suffix = [1]*n
        prefix = [1]*n


        for i in range(1,n):
            prefix[i] = prefix[i-1] * nums[i-1]


        for j in range(n-2, -1,-1):
            suffix[j] = prefix[j+1] * nums[j+1]




        for i in range(n):
            answer[i] = prefix[i]*suffix[i]

        return answer


```
