```python
class Solution:


    def frequency(self, s,t):

        hash = {}


        for char in s:

            hash[char] = hash.get(char,0) +1 
        
        for char in t:
            hash[char] = hash.get(char,0) -1
        

        for key, value in hash.items():
            
            if hash[key] != 0 :
                return False
        
        return True



```

```python
s="anagram"
t="nagaram"

sol = Solution()

sol.frequency(s,t)
```
