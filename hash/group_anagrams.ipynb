```python
class Solution:

    def group_anagrams(self,words):

        same = {}

        for word in words:

            new_word = "".join(sorted(word))

            if new_word not in same:
                same[new_word] = []
            
            same[new_word].append(word)

           
        return same.values()
            



```

```python
## upgrade version
from collections import defaultdict

def group_anagrams(self, words):
    same = defaultdict(list)  # 빈 리스트 자동 생성
    
    for word in words:
        same["".join(sorted(word))].append(word)
    
    return same.values()
```

```python
sol = Solution()
strs = ["eat","tea","tan","ate","nat","bat"]
sol.group_anagrams(strs)
```
