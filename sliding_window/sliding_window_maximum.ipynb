```python
# 최댓값 인덱스만 남김 -> 최적화를 위해서


# 시간복잡도 O(n)

from collections import deque

def sliding_window_maximum_opt(nums, k):
    dq = deque()  # 인덱스를 저장, 최댓값이 될가능성이 있는 인덱스만 남겨둠 
    result = []

    for i, num in enumerate(nums): 

         # 1. 뒤에서: 덱의 마지막 값이 현재 값보다 작으면 제거

        while dq and nums[dq[-1]] < nums[i]: 
              dq.pop()



          # 2. 현재 인덱스를 덱 뒤에 추가

        dq.append(i)



            # 3. 앞에서: 덱의 첫 값이 윈도우 범위를 벗어났으면 제거
        
        if dq[0] <= i-k : 
             dq.popleft()
        

        if i >= k-1: 
            result.append(nums[dq[0]])


    return result
            




        

        
```
