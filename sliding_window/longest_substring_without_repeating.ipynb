```python
class Solution:

    def length_of_longest_substring(self, s: str) -> int:


        left, right = 0,0
        max_length = 0
        hash = set()

        while right < len(s):

            length =  right - left +1

            if s[right] not in hash:
                hash.add(s[right])

                right +=1
                
                max_length = max(length, max_length)
            else:

                
                hash.remove(s[left])
                left +=1




        return max_length
```
