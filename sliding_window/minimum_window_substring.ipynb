```python
def minimum_window_substring(s, t) :
    
    need = Counter(t)
    left, right = 0,0
    count = {}
    # missing 추가 - 아직 부족한 총 문자 개수(중복포함)
    missing = len(t)
    start, end = 0,float("inf")




# n번실행 
    while right < len(s) :   
        count[s[right]] = count.get(s[right], 0) +1
        char = s[right]


        if char in need and count[char] <= need[char]:  # 부족분을 확인함 
            missing -=1 # gksrof
        
# left가 증가만 하고 감소안함 -> 추가적으로 n이 있지는 않음 
        while missing == 0 : # if missing !=0 이면 깨짐 ->현재 요구조건의 문자는 다 만족 

            if  (right-left) + 1 < (end-start):
                 # 지금 윈도우 길이 < 지금까지 최선의 답의 길이 - 지금만든 윈도우가 지금까지 찾은 최선의 답보다 더짧은가? 
               start, end = left, right+1   # 답갱신 

# 왼쪽 좁히기 
            left_char = s[left] # 왼쪽문자를 윈도에서 제거 준비
            count[left_char] -=1  # 카운트 감소 

            if left_char in need and count[left_char] < need[left_char]: # 실제 요건에서 부족해 졌는지 확인
                  missing +=1 
            left +=1# left ㅎㅏ
        right +=1



    return "" if end == float("inf") else s[start:end]




        
        
        

```
