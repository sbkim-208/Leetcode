```python
from collections import Counter

class Solution:
    def checkInclusion(self, s1: str, s2: str) -> bool:

        hash = Counter(s1)

        right, left = 0,0
        count = {}

        while right< len(s2):

            count[s2[right]] = count.get(s2[right], 0) +1
            

            if right - left + 1 == len(s1):

                if count == hash:
                    return True
                
                else:
                    count[s2[left]] -=1
                    if count[s2[left]] == 0:
                        del count[s2[left]]
                    left +=1
                
            right +=1
        
        return False
        


```
