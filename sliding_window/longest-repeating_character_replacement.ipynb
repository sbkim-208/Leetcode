```python
class Solution:
    def characterReplacement(self, s: str, k: int) -> int:
        hash = {}

        left, right = 0, 0
        max_length = 0

        while right < len(s):

            hash[s[right]] = hash.get(s[right], 0) +1
            length = right -left + 1 

            if length  - max(hash.values())<= k:

                
                
                max_length = max(length, max_length)
                right +=1

            else:
                hash[s[left]] -=1
                if hash[s[left]] ==0:
                    del hash[s[left]]

                left +=1
        
        return max_length



```
