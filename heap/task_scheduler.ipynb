```python
#Time: O(N log N)
#Space: O(N)
```

```python
from collections import Counter, deque
import heapq
from typing import List


class Solution:
    def leastInterval(self, tasks: List[str], n: int) -> int:



        tasks = Counter(tasks)

        heap = [-val for val in tasks.values()]
        heapq.heapify(heap)
        queue = deque() #  Queue to store tasks in cooldown: (remaining count, available time)

        time = 0

        while heap or queue: 
            time +=1

            if heap: 
                cnt = heapq.heappop(heap) + 1 # (executed once, 2 remaining)

                if cnt < 0 : #  cnt == 0 means task is fully done

                    queue.append((cnt, time + n))
            
            if queue and queue[0][1] == time:  #  If cooldown is over, push task back to heap
                heapq.heappush(heap, queue.popleft()[0])
        return time 
```

```python
solution = Solution()

tests = [
    (["A","A","A","B","B","B"], 2, 8),
    (["A","A","A","B","B","B"], 0, 6),
    (["A","A","A","A","A","A","B","C","D","E","F","G"], 2, 16),
    (["A"], 0, 1),
    (["A","B","C"], 2, 3),
]

for tasks, n, expected in tests:
    result = solution.leastInterval(tasks, n)
    status = "✅" if result == expected else "❌"
    print(f"{status} tasks={tasks}, n={n} | expected={expected}, got={result}")
```
