```python
import heapq
```

```python
class Solution: 

    def kth_origin(self,points, k): 

        heap = []

        for point in points: 

            x,y = point

            dist = x**2 + y**2

            heapq.heappush(heap, (-dist, point))

        while  len(heap) > k: 
            heapq.heappop(heap)
        

        return [point for dist, point in heap]
```

```python
sol = Solution()
points = [[1,3],[-2,2]]
k = 2
sol.kth_origin(points , k )
```
