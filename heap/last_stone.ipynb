```python
import heapq


def last_stone(stones): 
    heap = []

    for stone in stones: 

        heapq.heappush( heap, -stone)

 

    while len(heap) > 1 : 
        

        val1 = -heapq.heappop(heap)
        val2 = -heapq.heappop(heap)

        if val1 != val2:
             diff = val1 - val2
             heapq.heappush(heap, -diff)
    
    if len(heap) == 1: 
        return -heap[0]
    else: 
        return 0



    
    



#


```

```python
stones = [2, 7, 4, 1, 8, 1]

last_stone(stones)
```
