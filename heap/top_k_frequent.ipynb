```python
from collections import Counter

import heapq 



class Solution:

    def top_k_frequent(self, nums, k):


         nums = Counter(nums)

         heap = []

         for key, value in nums.items():
              heapq.heappush(heap, [value, key])

              if len(heap) > k:
                   heapq.heappop(heap)
        

         return [key for fre, key in heap]







```
