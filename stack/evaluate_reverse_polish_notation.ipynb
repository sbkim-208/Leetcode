```python
def evalRPN(tokens):
    stack = []
    operators = {'+', '-', '*', '/'}
    
    for token in tokens:
        if token not in operators:
            stack.append(int(token))
        else:
            b = stack.pop()
            a = stack.pop()
            if token == '+':
                stack.append(a + b)
            elif token == '-':
                stack.append(a - b)
            elif token == '*':
                stack.append(a * b)
            elif token == '/':
                # 0을 향해 버림(truncate toward zero)
                stack.append(int(a / b))
    
    return stack[0]

# 예시
print(evalRPN(["2","1","+","3","*"]))  # 9
print(evalRPN(["4","13","5","/","+"])) # 6
```
