```python
class Solution:


    def valid_parentheses_1(self, s:str):

        hash = {"}":"{", "]": "[", ")": "("}
        stack = []


        for char in s:

            if char not in hash:
                stack.append(char)
            

            elif char in hash:

                if not stack  or hash[char] != stack.pop():
                    return False
        
        return len(stack) == 0



```

```python
class Solution:

    def valid_parentheses_2(self,s:str):
        hash = {"}":"{", "]": "[", ")": "("}
        stack = []


        for char in s:
            

            if char in hash:

                if not stack:
                    return False
            
            top = stack.pop()

            if top != hash[char]:
                return False
        
        return len(stack) == 0
```

```python
## time complexity : O(n)

"""

I'll use a stack and a  hash map to solve this problem 

The hash map stores the matching pairs of closing and opening brackets 
for example, ')' maps to '('.

The stack is useful here because we need to check the most recently seen 
opening bracket when we encounter a closing bracket.
That's a Last-In-First-Out structure,
which is exactly what a stack does.


The time complexity is O(n), since we iterate through each character once, 
and the space is O(n) for the stack and hash map.

"""
```
