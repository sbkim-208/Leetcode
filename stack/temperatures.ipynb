```python
class Solution:

    def dailyTemperatures(self, temperatures):  # ← s 추가

        stack = []
        outcome = [0] * len(temperatures)

        for idx, temp in enumerate(temperatures):

            while stack and temperatures[stack[-1]] < temp:

                prev_idx = stack.pop()
                diff = idx - prev_idx
                outcome[prev_idx] = diff

            stack.append(idx)
        
        return outcome
```

```python
temperatures = [73, 74, 75, 71, 69, 72, 76, 73]


sol = Solution()
sol.dailyTemperatures(temperatures)
```
