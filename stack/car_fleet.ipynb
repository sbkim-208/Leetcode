```python
class Solution:

    def carFleet(self, target, position, speed):

        cars = sorted(zip(position,speed), reverse = True)

        fleet = 0
        times = []


        for car in cars:

            time = (target - car[0]) / car[1]

            if not times or time > times[-1]:
                fleet +=1

            times.append(time)

        return fleet
```

```python
position = [10, 8, 0, 5, 3]
speed    = [2,  4, 1, 1, 3]


sol = Solution()

sol.carFleet(12,position, speed)
```
